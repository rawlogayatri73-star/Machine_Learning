```python
df
x=df[['Selling_Price']]
y=df['Present_Price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_test,y_test,x_train,y_train
```

```python
m=LinearRegression()
m.fit(x_train,y_train)
```

```python
predict=m.predict(x_test)
```

```python
result = pd.DataFrame({'actual':y_test, 'predicted':predict})
```

```python
result.tail()
```

```python
mae=mean_absolute_error(y_test,predict)
print("Mean Absolute Error:",mae)
print("Mean Squared Error:",mean_squared_error(y_test,predict))
print("Root Mean Squared Error:",np.sqrt(mean_squared_error(y_test,predict)))
print("R2 Score:",r2_score(y_test,predict))
```

```python
plt.scatter(x,y)
plt.xlabel('Present_Price')
plt.ylabel('selling_Price')
plt.plot(x_train,m.predict(x_train),color = 'red')
```

```python
outlier_cols = ["Selling_Price", "Present_Price"]  # replace with your column names

mask = pd.Series(True, index=df.index)

for col in outlier_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    mask &= df[col].between(q1 - 1.5 * iqr, q3 + 1.5 * iqr)

df_clean = df[mask].copy()

print("Before:", df.shape)
print("After: ", df_clean.shape)
```
